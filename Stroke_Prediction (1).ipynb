```python
df["bmi"].fillna(df["bmi"].mode()[0],inplace=True)
df.isna().sum()
```

```python
df.dtypes
```

```python
sns.lineplot(x="age",y="hypertension",data=df)
plt.title("Line plot")
plt.xlabel("Age")
plt.ylabel("Outcome")
plt.show()
```

```python
sns.barplot(x="age",y="smoking_status",data=df)
plt.title("Bar plot")
plt.xlabel("Age")
plt.ylabel("Outcome")
plt.show()
```

```python
sns.boxplot(x="age",y="ever_married",data=df)
plt.title("Box plot")
plt.xlabel("Age")
plt.ylabel("Outcome")
plt.show()
```

```python
sns.pairplot(df)
plt.title("Pair Plot")
plt.show()
```

```python
sns.scatterplot(x="age",y="avg_glucose_level",data=df)
plt.title("Scatter Plot")
plt.xlabel("Age")
plt.ylabel("Outcome")
plt.show()
```

```python
sns.histplot(x='avg_glucose_level',data=df)
plt.title('Histogram Plot')
plt.xlabel('Insulin')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.pie(df['hypertension'].value_counts(),labels=df['hypertension'].unique(),autopct='%1.1f%%')
plt.title('Pie Chart')
plt.show()
```

```python
sns.heatmap(df.select_dtypes(include=[np.number]).corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```
