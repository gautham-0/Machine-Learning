```python
for col in ['2022', '2023', '2024']:
    if col in df.columns:
        df[col] = df[col].fillna(df[col].mean())

display(df.isnull().sum())
```

```python
df['Battery_power_mAh'] = df['Battery_power_mAh'].str.replace(' mAh', '', regex=False).astype(int)
display(df['Battery_power_mAh'].head())
```

```python
df.dtypes
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
sns.lineplot(x='price_range',y='Bluetooh',data=df)
plt.title('Line Plot')
plt.xlabel('Price_Range')
plt.ylabel('Bluetooth')
plt.show()
```

```python
sns.boxplot(x="price_range",y='Primary_camera',data=df)
plt.title('Box Plot')
plt.xlabel("Price_Range")
plt.ylabel("Primary_camera")
plt.show()
```

```python
counts = df["Bluetooh"].value_counts()
plt.pie(counts,labels=counts.index,autopct="%1.1f%%")
plt.title("Bluetooh")
plt.show()
```

```python
plt.hist(df["Front_camera"],bins=10)
plt.title("Distribution of Front_camera")
plt.xlabel("Front_camera")
plt.ylabel("price_range")
plt.show()
```

```python
sns.scatterplot(
    x="Ram_mb",
    y="price_range",
    data=df
)
plt.title("Battery_power_mAh vs price_range")
plt.xlabel("Battery_power_mAh")
plt.ylabel("price_range")
plt.show()
```

```python
sns.pairplot(df)
plt.show()
```

```python
sns.heatmap(df.corr(numeric_only=True), annot=True)
plt.title("Correlation Matrix")
plt.show()
```
